```python
from app.retrieval.retriever import retrieve, get_relevant_documents
```

```python
query_example = "What is the input parameters for createMOI"
# filters_example = {"release": "18", "series": "38"}

relevant_chunks = retrieve(
    query=query_example,
    top_k=5,
    # filters=filters_example,
)
```

```python
print(relevant_chunks)
```

```python
print(f"\nTop {len(relevant_chunks)} results for: '{query_example}'\n")
for i, res in enumerate(relevant_chunks, 1):
    print(f"[{i}] score = {res['score']:.4f}")
    print(f"release: {res['metadata'].get('release')}")
    print(f"series:  {res['metadata'].get('series')}")
    print(f"spec:    {res['metadata'].get('spec')}")
    print(f"text:\n{res['text']}...\n")
    print("-" * 80)
```
